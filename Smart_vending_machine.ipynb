product class

```python
class Product:
    def __init__(self, product_id, name, price, stock):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.stock = stock

    def display(self):
        print(f"ID: {self.product_id}")
        print(f"Name: {self.name}")
        print(f"Price: ₹{self.price}")
        print(f"Stock: {self.stock}")
```

```python
p1 = Product(101, "Coke", 40, 10)
p2= Product(102, "Pepsi", 40, 60)
p3= Product(103, "Lays", 20, 40)
p4= Product(104, "Kurkure", 20, 40)
p5= Product(105, "Oreo", 20, 35)
p6= Product(106, "Dairy Milk", 30, 50)

p1.display()
p2.display()
p3.display()
p4.display()
p5.display()
p6.display()
```

```python
#Registration of Customers

class Customer:
    def __init__(self, customer_id, name):
        self.customer_id = customer_id
        self.name = name
        self.reward_points = 0
        self.history = []

    def display_customer(self):
        print("Customer ID:", self.customer_id)
        print("Name:", self.name)
        print("Reward Points:", self.reward_points)
```

Vendingmachine class

```python
class VendingMachine:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def show_products(self):
        for product in self.products:
            print(
                product.product_id,
                product.name,
                "₹" + str(product.price),
                "Stock:", product.stock
            )
```

```python
vm = VendingMachine()

vm.add_product(Product(1, "Coke", 40, 10))
vm.add_product(Product(2, "Chips", 20, 15))
vm.add_product(Product(3, "Juice", 35, 8))

vm.show_products()
```
